# Train YOLO

Notebook thực hiện fine-tuning mô hình **pretrained** trên một bộ dữ liệu đã được chuẩn hóa theo định dạng YOLO và chia thành hai tập:

* **Train:** dùng để huấn luyện mô hình.
* **Validation:** dùng để theo dõi quá trình huấn luyện và lựa chọn mô hình tốt nhất.

Quá trình huấn luyện ưu tiên sử dụng các thiết lập mặc định của Ultralytics để tạo baseline đơn giản, ổn định và dễ tái lập. Kết quả tốt nhất (`best.pt`/`best.onnx`) từ `/runs/detect/train/weights/` sẽ được sử dụng cho các bước tiếp theo của hệ thống.

```python
%pip install -q -U ultralytics

from ultralytics import YOLO

model = YOLO("/kaggle/input/datasets/nvbangg/yolo-drowning-dataset/best-yolo.pt")
model.train(
    data="/kaggle/input/datasets/nvbangg/yolo-drowning-dataset/data.yaml",
    cls_remap=True, # ánh xạ trọng số theo tên class khi thứ tự ID khác nhau
    resume=False, # fine-tune mới từ best-yolo.pt
    epochs=100,
    patience=10, # dừng khi 10 epochs liên tiếp không cập nhật best.pt
    batch=64, # để tận dụng vRam
    device=[0, 1], # dùng 2 GPU
    deterministic=False, # chạy nhanh hơn nhưng không đảm bảo các lần chạy giống nhau
)

YOLO(model.trainer.best).export(format="onnx", opset=17)
```
